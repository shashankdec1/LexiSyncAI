## Inspect legal document structure
This cell loads the page inventory created in the previous notebook and scans the extracted text for likely section headings. For the GDPR, it looks for lines beginning with “Article” followed by a number. For the EDPB guidelines, it looks for chapter headings and numbered subsections.

The notebook prints each matching heading with its PDF page number, up to 30 matches per document. This gives a quick overview of the documents’ structure and helps locate sections for later analysis. The patterns are simple text checks, so they may also match references to articles in regular paragraphs.

```python
from pathlib import Path
import csv
import re

project_dir = Path.home() / "Desktop/LexiSyncAI"
inventory_path = project_dir / "data/processed/page_inventory.csv"

with inventory_path.open(encoding="utf-8", newline="") as file:
    pages = list(csv.DictReader(file))

patterns = {
    "gdpr_regulation.pdf": re.compile(r"^\s*Article\s+\d+", re.IGNORECASE),
    "edpb_fines_guidelines.pdf": re.compile(
        r"^\s*(?:CHAPTER\s+\d+|\d+\.\d+\s*[-–])",
        re.IGNORECASE,
    ),
}

for document, pattern in patterns.items():
    print(f"\n=== {document} ===")
    matches = 0

    for page in pages:
        if page["document"] != document:
            continue

        for line in page["text"].splitlines():
            if pattern.search(line):
                print(f"PDF page {page['page_number']}: {line.strip()}")
                matches += 1
                if matches >= 30:
                    break

        if matches >= 30:
            break
```
